# Aging mouse brain: reproduce the Figure 5A local maps

This tutorial runs mismatch-aware local spatial differential expression for **Gamt** in five aligned aging mouse-brain MERFISH sections. The package includes the raw counts, original and aligned coordinates, and cell-type labels needed for the 4.3-month reference and the 6.6-, 15.8-, 30.9-, and 34.5-month queries.

The settings match the Figure 5A analysis: raw counts are not log-transformed and are normalized to a library size of 250, the density channel receives 25% of the standardized mismatch-feature energy, the gene-specific global offset and local covariates for library size and detection rate are included, mismatch-aware variance adjustment and connected-region cleanup are enabled, and cell-type adjustment is disabled. Grid-level significance uses $q \leq 0.05$.

```python
import spAlignDE

WORKFLOW_SEED = 1
seed_controls = spAlignDE.set_random_seed(
    WORKFLOW_SEED,
    deterministic_torch=True,
)

from spAlignDE.datasets import (
    AGING_BRAIN_FIGURE5A_REFERENCE,
    AGING_BRAIN_FIGURE5A_SAMPLES,
    aging_brain_figure5a_genes,
    aging_brain_figure5a_metadata,
    load_aging_brain_figure5a,
)
from spAlignDE import fit_local_de, plot_local_result, prepare_inference
```

## Load the packaged manuscript subset

All 300 expression columns are retained because mismatch-risk estimation screens a broad stable-gene candidate panel even though only `Gamt` is fitted below.

```python
metadata = aging_brain_figure5a_metadata()
risk_genes = list(aging_brain_figure5a_genes())
data = load_aging_brain_figure5a()

print(f"Samples: {AGING_BRAIN_FIGURE5A_SAMPLES}")
print(f"Cells: {len(data):,}; risk-gene candidates: {len(risk_genes)}")
data.groupby("sample_id", sort=False).size().rename("n_cells").to_frame()
```

## Construct the shared grid and mismatch-risk maps

The aligned coordinates define the common inference space. With `grid_n=None`, the package applies its data-driven shared-grid rule.

```python
prepared = prepare_inference(
    data,
    reference=AGING_BRAIN_FIGURE5A_REFERENCE,
    genes=["Gamt"],
    risk_genes=risk_genes,
    density_energy_share=0.25,
    library_size=250,
    grid_n=None,
    n_jobs=8,
    random_state=1,
)

{
    "grid_n": prepared.metadata["grid_n"],
    "shared_grid_locations": prepared.metadata["shared_grid_locations"],
    "grid_selection": prepared.metadata["grid_n_source"],
}
```

## Fit the four age-versus-reference contrasts

The result uses mismatch-aware local variance adjustment. `region_cleanup=True` affects only the reported connected-region mask; local statistics, P values, and q-values are unchanged.

```python
result = fit_local_de(
    prepared,
    genes=["Gamt"],
    contrast="vs_reference",
    alpha=0.05,
    mismatch_aware=True,
    technical_adjustment=True,
    cell_type_adjustment=False,
    global_offset=True,
    region_cleanup=True,
    n_jobs=3,
    random_state=None,
)
```

## Display the Figure 5A local-statistic maps

Each row is one age-versus-4.3-month contrast. Red contours mark the connected FDR-significant region reported by the fitted result.

```python
figure = plot_local_result(
    result,
    "Gamt",
    show_expression=False,
)
```
